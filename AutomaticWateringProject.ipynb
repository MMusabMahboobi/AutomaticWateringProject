# Automatic Watering Project

**19/08/2026, 10:03 AM:**
This log will firstly explain the core experiment I plan to perform and secondly explain some of my hardware choices.

### The Experiment
I will be comparing the amount of water used by an automatic watering system that waters a plant whenever the soil moisture falls below a certain threshold vs manual watering at a set time every day.

Independent Variable: Watering method; fixed-time manual watering vs. dryness-triggered automatic watering

Dependent Variables:
* Amount of water used in two weeks
* Moisture over time

Control Variables:
* Plant Type: Undecided as of yet, but they will be of the same size and species
* Location: Level of sunlight and humidity will be the same
* Type of Water: Simple tap water for both plants
* Soil Type
* Soil Starting Moisture
* Pot Type and Design

I will record any unusual events such as spills, power outages (common in Pakistan), or reservoir refills

### Hardware Choices
* Arduino Uno: A simple computer that is great for this project. It will read the sensors and control the pump accordingly
* Capacitive Soil Moisture Sensors Over Resistive Ones: This is for two main reasons:
    1. Resistive soil moisture sensors have exposed metal, which corrodes over time
    2. Resistive soil moisture sensors give false readings from salts that are often present in soils
* Peristaltic Pump: I chose this type of pump because the fluid it pumps remains inside of a tube and never touches any mechanical parts that might be ruined
* MOSFET Driver Over Relay: A relay has mechanical parts which may fail or wear down over time, whereas a MOSFET driver switches using electricity. A MOSFET drive also provides faster switching
* Capacitors: These will be used to reduce electrical noise caused by the pump
* Flyback Diode: When switched off, the pump may produce a massive voltage spike. The flyback diode will prevent this voltage spike from damaging my Arduino Uno
